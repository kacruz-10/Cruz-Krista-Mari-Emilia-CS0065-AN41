1. Create a knowledge base

```python
facts = {
    "temperature": 34,
    "is_raining": True,
    "humidity": 90,
    "smoke_detected": False,
    "occupancy": True
}
```

2. Create at least 3 IF-THEN rules

```python
def rule_hot(kb):
    if kb.get("temperature", 0) > 30:
        return "Turn on the aircon"

def rule_rain(kb):
    if kb.get("is_raining", False):
        return "Bring umbrella"

def rule_smoke(kb):
    if kb.get("smoke_detected", False):
        return "Trigger fire alarm"

rules = [rule_hot, rule_rain, rule_smoke]
```

3. Implement an inference engine

```python
def run_inference_engine(kb, rule_base):
    inferred_actions = []
    for rule in rule_base:
        action = rule(kb)
        if action:
            inferred_actions.append(action)
    return inferred_actions
```

4. Build a case base

```python
case_base = [
    {
        "problem": {"won_t_turn_on": True, "strange_noise": False, "indicator_light_off": True},
        "solution": "Check the power cable and outlet connection."
    },
    {
        "problem": {"won_t_turn_on": False, "strange_noise": True, "indicator_light_off": False},
        "solution": "Clear obstructions from the motor and check bearings."
    },
    {
        "problem": {"won_t_turn_on": True, "strange_noise": True, "indicator_light_off": True},
        "solution": "Perform a hard reset and inspect the internal fuse."
    }
]
```

5. Retrieve and reuse step

```python
def retrieve_and_reuse(new_problem, cases):
    print("\n[RETRIEVE & REUSE]")
    best_match_solution = None
    max_similarity = -1

    for case in cases:
        # calculate overlap similarity matching based on matching feature values
        past_problem = case["problem"]
        match_count = sum(1 for k, v in new_problem.items() if past_problem.get(k) == v)

        if match_count > max_similarity:
            max_similarity = match_count
            best_match_solution = case["solution"]

    print("Matching problem found in Case Base.")
    print(f"Retrieved Solution: {best_match_solution}")
    return best_match_solution
```

6. Revise step

```python
def revise_solution(solution):
    # allow adjustment or return unchanged
    print(f"\n[REVISE] Retrieved Solution: {solution}")
    choice = input("Do you want to adjust this solution? (y/n): ").strip().lower()
    if choice == 'y':
        return input("Enter revised solution: ").strip()
    return solution
```

7. Retain step

```python
def retain_case(new_problem, final_solution, cases):
    new_case = {"problem": new_problem, "solution": final_solution}
    cases.append(new_case)
    print("[RETAIN] New case successfully added to the case base.")
```

8. Display output

```python
if __name__ == "__main__":
    print("=== Rule-Based Reasoning (RBR) Output ===")
    actions = run_inference_engine(facts, rules)
    print("Rule-Based Reasoning Actions:")
    for action in actions:
        print(f"- {action}")

    print("\n=== Case-Based Reasoning (CBR) Output ===")
    current_problem = {"won_t_turn_on": True, "strange_noise": False, "indicator_light_off": True}

    retrieved_solution = retrieve_and_reuse(current_problem, case_base)

    final_solution = revise_solution(retrieved_solution)
    retain_case(current_problem, final_solution, case_base)

    print(f"\nTotal cases in Case Base now: {len(case_base)}")
```
