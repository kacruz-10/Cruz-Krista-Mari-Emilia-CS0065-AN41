```python
import math
```

1. Initialize the case base with past disaster records

```python
case_base = [
    {
        "name": "Typhoon Yolanda (2013)",
        "problem": {"type": "typhoon", "area": "coastal", "population": 400000},
        "solution": {"strategy": "Mass coastal evacuation to reinforced centers", "evacuee_percent": 0.50, "shelter_capacity": 500}
    },
    {
        "name": "Typhoon Rolly (2020)",
        "problem": {"type": "typhoon", "area": "rural", "population": 150000},
        "solution": {"strategy": "Evacuate agricultural zones and secure livestock", "evacuee_percent": 0.35, "shelter_capacity": 300}
    },
    {
        "name": "Marikina River Overflow (2020)",
        "problem": {"type": "flood", "area": "urban", "population": 220000},
        "solution": {"strategy": "Riverside evacuation using local rescue boats", "evacuee_percent": 0.25, "shelter_capacity": 350}
    },
    {
        "name": "Batangas Earthquake (2023)",
        "problem": {"type": "earthquake", "area": "rural", "population": 90000},
        "solution": {"strategy": "Open-field assembly and temporary tent setup", "evacuee_percent": 0.40, "shelter_capacity": 200}
    }
]
```

2. Define new problem

```python
new_problem = {
    "type": "typhoon",
    "area": "urban",
    "population": 180000
}
```

3. Calculate similarity score

```python
def similarity(new, old):
    score = 0.0
    if new["type"] == old["type"]:
        score += 0.5
    if new["area"] == old["area"]:
        score += 0.3
    score += 0.2 * (min(new["population"], old["population"]) / max(new["population"], old["population"]))
    return score
```

4. Adapt past solution to new problem

```python
def adapt(new, old_case):
    old = old_case["solution"]
    evacuees = round(new["population"] * old["evacuee_percent"])
    shelters = math.ceil(evacuees / old["shelter_capacity"])
    strategy = old["strategy"]
    if new["area"] == "urban":
        strategy += "; add vertical evacuation and multi-story deployment"
    return {
        "strategy": strategy,
        "evacuee_percent": old["evacuee_percent"],
        "shelter_capacity": old["shelter_capacity"],
        "evacuees": evacuees,
        "shelters needed": shelters
    }

def main():
    print("=== CBR DISASTER RESPONSE SYSTEM ===")
    print(f"\nNEW PROBLEM: Typhoon with heavy rain in an urban area")
    print(f"Type: {new_problem['type']}, Area: {new_problem['area']}, Population: {new_problem['population']:,}")
```

5. Retrieve most similar case

```python
    print("\nSIMILARITY SCORES:")
    ranked = sorted(case_base, key=lambda c: similarity(new_problem, c["problem"]), reverse=True)
    for i, c in enumerate(ranked, 1):
        print(f" {i}. {c['name']:<32} {similarity(new_problem, c['problem']):.2f}")

    best = ranked[0]
    print(f"\nMOST SIMILAR CASE: {best['name']}")
    print(f"Old strategy: {best['solution']['strategy']}")
```

6. Reuse and adapt solution

```python
    final = adapt(new_problem, best)
    print("\nADAPTED FINAL SOLUTION:")
    print(f"Strategy: {final['strategy']}")
    print(f"Evacuees: {final['evacuees']:,} ({final['evacuee_percent']:.0%} of population)")
    print(f"Shelters: {final['shelters needed']:,} (capacity {final['shelter_capacity']} each)")
```

7. Retain new

```python
    case_base.append({"name": "Typhoon Urban Event (2026)", "problem": new_problem, "solution": final})
    print("\nNew case saved!")
    print("\nUPDATED CASE BASE:")
    for i, c in enumerate(case_base, 1):
        print(f" {i}. {c['name']}")

if __name__ == "__main__":
  main()
```
