```python
class VacuumAgent:
    def __init__(self, start_room="A"):
        self.location = start_room

    def perceive_and_act(self, environment):
        """
        Rule 1: If the current room is dirty, clean it.
        Rule 2: If the current room is clean, move to the other room.
        """
        action_taken = ""
        if environment[self.location] == "Dirty":
            environment[self.location] = "Clean"
            action_taken = "Cleaned"
        else:
            self.move(environment)
            action_taken = "Moved"
        return action_taken

    def move(self, environment):
        """Switch rooms between A and B."""
        rooms = list(environment.keys())
        if len(rooms) == 2:
            self.location = rooms[1] if self.location == rooms[0] else rooms[0]
        else:
            current_idx = rooms.index(self.location)
            self.location = rooms[(current_idx + 1) % len(rooms)]


def run_simulation():
    # set up the environment
    environment = {"A": "Clean", "B": "Clean"}

    print("--- Environment Setup ---")
    for room in environment:
        state = input(
            f"Enter initial state for Room {room} (Dirty/Clean): "
        ).strip().capitalize()
        if state in ["Dirty", "Clean"]:
            environment[room] = state
        else:
            print("Invalid input defaulting to Clean.")
            environment[room] = "Clean"

    agent = VacuumAgent(start_room="A")

    try:
        steps = int(input("How many steps should the simulation run? "))
    except ValueError:
        steps = 5

    print("\n--- Simulation Started ---")
    for step in range(1, steps + 1):
        print(f"\nStep {step}:")
        print(f"  - Agent's Current Location: Room {agent.location}")

        action = agent.perceive_and_act(environment)
        print(f"  - Action Taken: {action}")

        env_str = ", ".join([f"{room} = {state}" for room, state in environment.items()])
        print(f"  - Current State of Environment: {env_str}")

    # check if all rooms are clean
    all_clean = all(state == "Clean" for state in environment.values())
    if all_clean:
        print("\nAgent is done making rounds (all rooms are clean).")
    else:
        print("\nAgent finished steps, but dirty rooms still remain.")


if __name__ == "__main__":
    run_simulation()
```

```python
import random

class VacuumAgent:
    def __init__(self, start_room="A"):
        self.location = start_room

    def perceive_and_act(self, environment):
        """
        Bonus Task Logic:
        - If the current room is dirty, clean it.
        - If clean, randomly choose one of the remaining dirty rooms to clean next.
        - If all rooms are clean, move randomly to another room.
        """
        action_taken = ""
        dirty_rooms = [room for room, state in environment.items() if state == "Dirty"]

        if environment[self.location] == "Dirty":
            environment[self.location] = "Clean"
            action_taken = f"Cleaned Room {self.location}"
        elif dirty_rooms:
            # Randomly choose which dirty room to clean next
            target_room = random.choice(dirty_rooms)
            self.location = target_room
            environment[self.location] = "Clean"
            action_taken = f"Moved to and Cleaned Room {target_room}"
        else:
            # If all rooms are clean, move randomly between rooms
            all_rooms = list(environment.keys())
            all_rooms.remove(self.location)
            self.location = random.choice(all_rooms)
            action_taken = f"All rooms clean. Moved to Room {self.location}"

        return action_taken


def visualize_grid(environment, agent_location):
    """Visualize rooms and agent position using a text-based grid."""
    print("+---------+---------+---------+")
    print("| Room A  | Room B  | Room C  |")
    print(f"|St:{environment['A']:<5}|St:{environment['B']:<5}|St:{environment['C']:<5}|")

    # Position indicator for the agent
    pos_a = "[AGENT] " if agent_location == "A" else "        "
    pos_b = "[AGENT] " if agent_location == "B" else "        "
    pos_c = "[AGENT] " if agent_location == "C" else "        "
    print(f"|{pos_a}|{pos_b}|{pos_c}|")
    print("+---------+---------+---------+")


def run_simulation():
    # Set up three rooms (A, B, and C) for the Bonus Task[cite: 1]
    environment = {"A": "Clean", "B": "Clean", "C": "Clean"}

    print("--- Environment Setup (Bonus Task: 3 Rooms) ---")
    for room in environment:
        state = input(f"Enter initial state for Room {room} (Dirty/Clean): ").strip().capitalize()
        if state in ["Dirty", "Clean"]:
            environment[room] = state
        else:
            print("Invalid input defaulting to Clean.")
            environment[room] = "Clean"

    agent = VacuumAgent(start_room="A")

    try:
        steps = int(input("How many steps should the simulation run? "))
    except ValueError:
        steps = 5

    print("\n--- Simulation Started ---")
    for step in range(1, steps + 1):
        print(f"\nStep {step}:")
        print(f"  - Agent's Current Location: Room {agent.location}")

        action = agent.perceive_and_act(environment)
        print(f"  - Action Taken: {action}")

        env_str = ", ".join([f"{room} = {state}" for room, state in environment.items()])
        print(f"  - Current State of Environment: {env_str}")

        print("  - Grid Visualization:")
        visualize_grid(environment, agent.location)

    # Check if all rooms are clean after simulation steps finish
    all_clean = all(state == "Clean" for state in environment.values())
    if all_clean:
        print("\nAgent is done making rounds (all rooms are clean).")
    else:
        print("\nAgent finished steps, but dirty rooms still remain.")


if __name__ == "__main__":
    run_simulation()
```
